```python
import pandas as pd

df = pd.read_csv("data/domains.csv")

# Availability = inactive
available = df[df['status'].str.endswith("inactive")]
print("Likely available domains:", len(available))

# Distribution of statuses
print(df['status'].value_counts())
```

```python
available_by_zone = available.groupby("zone").size().sort_values(ascending=False)
available_by_zone = available_by_zone[available_by_zone <= 30] # Only print less than 30
available_by_zone = available_by_zone[available_by_zone != 10] # Ignore reserved digits
print(available_by_zone)
```

```python
def show_possible_by_zone(df, zone):
    possible_statuses = ["inactive", "undelegated", "unknown"]
    subset = df[
        (df["zone"] == zone) &
        (df["status"].str.split().str[-1].isin(possible_statuses))
    ]
    return subset[['domain','zone','status']]

print(show_possible_by_zone(df, "ki").to_string(index=False))
```

```python
# Filter marketed, priced, premium domains
marketed_statuses = df[df['status'].str.contains("marketed|priced|premium", case=False)]

# Show detail preview if you want
print(marketed_statuses[['domain', 'zone', 'status']].head(20).to_string(index=False))

# Group marketed ones by zone
marketed_by_zone = marketed_statuses.groupby("zone").size().sort_values(ascending=False)
print(marketed_by_zone)

# Optional: merge with available counts for comparison
zone_comparison = pd.DataFrame({
    "available": available.groupby("zone").size(),
    "marketed": marketed_by_zone
}).fillna(0).astype(int).sort_values("marketed", ascending=False)

print(zone_comparison.to_string())
```

```python
# Read and rank TLDs by price
tld_df = pd.read_csv("data/output.csv")

# Sort by registration_price (cheapest first)
tld_ranked = tld_df.sort_values('registration_price', ascending=True)

print("TLDs ranked by price (cheapest first):")
print(tld_ranked.to_string(index=False))

# Show top 10 cheapest
print("\nTop 10 cheapest TLDs:")
print(tld_ranked.head(10).to_string(index=False))
```
